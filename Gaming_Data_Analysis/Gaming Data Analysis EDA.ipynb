```python
small_missing_cols = []

for col in df.columns:
    if df[col].isnull().sum() <= 2:
        small_missing_cols.append(col)

df = df.dropna(subset=small_missing_cols)
```

```python
df.isnull().sum()
```

```python
missing_pct = (df[['user_rating','metacritic']].isnull().sum() / len(df) * 100)

missing_pct
```

```python
df[['user_rating', 'metacritic']].describe()
```

```python
df['user_rating'] = df['user_rating'].fillna(df['user_rating'].median())

df['metacritic'] = df['metacritic'].fillna(df['metacritic'].median())
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
num_cols = df.select_dtypes(include=['int64','float64']).columns

df[num_cols].describe().T
```

```python
df['release_year'] = df['release_year'].astype(int)
```

```python
drop_cols = [
    'serial_no',
    'cover_image_url',
    'official_website',
    'url_slug',
    'description_clean'
]

df=df.drop(columns=drop_cols)
```

```python
genre_table = (
                df[['game_id', 'all_genres']].copy()
)

genre_table['all_genres'] = (
                            genre_table['all_genres'].str.split('|')
)

genre_table = genre_table.explode('all_genres')

genre_table.rename(
                    columns={'all_genres': 'genre'},inplace=True
)

genre_table['genre'] = genre_table['genre'].str.strip()

genre_table.head()
```

```python
genre_table.shape
```

```python
genre_table['genre'].nunique()
```

```python
genre_table.duplicated().sum()
```

```python
platform_table = (
                    df[['game_id', 'all_platforms']].copy()
)

platform_table['all_platforms'] = (
                                    platform_table['all_platforms'].str.split('|')
)

platform_table = platform_table.explode('all_platforms')

platform_table.rename(
                        columns={'all_platforms':'platform'},inplace=True
)

platform_table['platform'] = (
                                platform_table['platform'].str.strip()
)

platform_table.drop_duplicates(inplace=True)
```

```python
tag_table = (
                df[['game_id', 'all_tags']].copy()
)

tag_table['all_tags'] = (
                            tag_table['all_tags'].str.split('|')
)

tag_table = tag_table.explode('all_tags')

tag_table.rename(
                    columns={'all_tags':'tag'},inplace=True
)

tag_table['tag'] = tag_table['tag'].str.strip()

tag_table.drop_duplicates(inplace=True)
```

```python
store_table = (
                df[['game_id', 'available_stores']].copy()
)

store_table['available_stores'] = (
                                    store_table['available_stores'].str.split('|')
)

store_table = store_table.explode('available_stores')

store_table.rename(
                    columns={'available_stores':'store'},inplace=True
)

store_table['store'] = store_table['store'].str.strip()

store_table.drop_duplicates(inplace=True)
```
